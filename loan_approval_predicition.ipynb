```python
# finding the outliers in the dataset

# using IQR method
# using for loop
# using concat() method

outliers = pd.DataFrame()
def check_outliers(df_loan1):
    global outliers  # Declare outliers as a global variable
    
    for i in df_loan1.columns:  # Only process numeric columns
        q1 = df_loan1[i].quantile(0.25)
        q3 = df_loan1[i].quantile(0.75)
        iqr = q3 - q1
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr
        outlier = df_loan1[(df_loan1[i] < lower_bound) | (df_loan1[i] > upper_bound)]
        print(f"Outliers in {i}:")
        outliers = pd.concat([outliers, outlier])
        print(outliers)
        print('\n')

columns = ["Age", "CreditScore", "Experience", "LoanDuration", "JobTenure", "BaseInterestRate", "InterestRate"]
check_outliers(df_loan[columns])


```

```python
# Remove outliers

# Align the indices of the outliers DataFrame with df_loan
outliers_indices = outliers.index.intersection(df_loan.index)
df_loan.drop(outliers_indices, inplace=True)

df_loan.shape
# Remove outliers
```

```python
# dropping the ApplicationDate column
# using drop() method
# and assigning it to df_loan
# checking the shape of the dataset
# using shape attribute
x= df_loan.drop(['ApplicationDate','LoanApproved'], axis=1)
y= df_loan['LoanApproved']
x.shape
y.shape
```

```python
#importing the StandardScaler class

from pandas import DataFrame
x_scaled=StandardScaler().fit_transform(x)
DataFrame(x_scaled)
```

```python
#splitting the dataset into training and testing sets
# using train_test_split() method
# from sklearn.model_selection module

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x_scaled,y,test_size=0.2,random_state=42)
```

```python
# checking the shape of the training and testing sets
# using shape attribute
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# importing the DecisionTreeClassifier class
from sklearn import tree
```

```python
# creating an object of DecisionTreeClassifier class
dec_tree= tree.DecisionTreeClassifier(criterion='entropy',max_depth=5,random_state=1)
```

```python
# fitting the model
dec_tree.fit(x_train,y_train) 
```

```python
# predicting the values
# using predict() method

Y_pred=dec_tree.predict(x_test)
```

```python
Y_pred
```

```python
# checking the accuracy of the model
# using accuracy_score() method
# using classification_report() method

from sklearn import metrics
```

```python
Accuracy= metrics.accuracy_score(y_test,Y_pred) # Calculating the accuracy of the model build
Recall= metrics.recall_score(y_test,Y_pred) # Calculating the recall of the model build
Precision= metrics.precision_score(y_test,Y_pred) # Calculating the precision of the model build
F1score=metrics.f1_score(y_test,Y_pred) # Calculating the F1 Score of the model build
print('Accuracy:', round(Accuracy,2))
print('Recall:', round(Recall,2))
print('Precision:',round(Precision,2))
print('F1 Score:',round(F1score,2))
```

```python
sns.heatmap(metrics.confusion_matrix(y_test,Y_pred),annot=True,fmt='d',cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```
